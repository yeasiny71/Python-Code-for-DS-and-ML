## if,elif, else Statements

```python
#some conditions, like if 1<2, then perform some action
if 1<2:
    print("yep, 1 is less than 2")
```

```python
#if the condition is true, then execute the block of code 
if True:
    print("It's true")
```

```python
#when a satisfied condition is met, then execute the block of code and exit the if statement
if 1 == 2:
    print("First")
elif 3 == 3:
    print("Second")

else:
    print("last")
```

### Map function

```python
seq = [1,2,3,4,5]
```

```python
map(times2, seq) #map function will apply times2 function to each element of seq 
```

```python
list(map(times2, seq)) #to see the output of map function, we need to convert it to list
```

### Lambda function 


```python
def times2(var):return var*2 #written in one line, works the same way
```

```python
times2(4) 
```

```python
#to convert into lambda, remove the func name and return statement, and add lambda keyword at the beginning
t = lambda var: var*2
t(6)
```

```python
#lambda pass_in : output
list(map(lambda var: var*2, seq)) #element of seq passed through lambda functon, lambda takes each element from seq and maps the var*2 output then arranged in a list
```

### Filter function

```python
#filter func should get a function or lambda function that returns a boolean value, and an iterable like list, filter will return only the elements that satisfy the condition defined in the function or lambda function

# lambda func is taking each value in seq and checking if even or not, for even it is returning true, otherwise false, filter will return only the values that are even in seq
list(filter(lambda num: num%2 == 0, seq))
```
