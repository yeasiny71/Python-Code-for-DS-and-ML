# Series
A Series is very similar to a NumPy array (in fact it is built on top of the NumPy array object). What differentiates the NumPy array from a Series, is that a Series can have axis labels, meaning it can be indexed by a label, instead of just a number location. It also doesn't need to hold numeric data, it can hold any arbitrary Python Object.


```python
import numpy as np
import pandas as pd
```

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array(my_list)
d ={'a':10, 'b':20, 'c':30}
```

```python
pd.Series(data=my_list)
```

```python
pd.Series(data=my_list, index=labels)
```

```python
pd.Series(my_list, labels)
```

```python
#passing numpy array inside series
pd.Series(arr)
```

```python
pd.Series(arr,labels)
```

```python
d
```

```python
#dictionary passed into series
pd.Series(d)  #keys are used as index and values are used as data
```

```python
#series holding string data 
pd.Series(labels)
```

```python
ser1 = pd.Series([1,2,3,4], ['USA','Germany','USSR','Japan'])
```

```python
ser1
```

```python
ser2 = pd.Series([1,2,5,4], ['USA','Germany','Italy','Japan'])
```

# Missing Data


```python
d = {'A':[1,2,np.nan], 'B':[5,np.nan,np.nan],'C':[1,2,3]}
```

```python
d
```

```python
df =pd.DataFrame(d)
```

```python
df
```

```python
#dropping missing values from a dataset
#will drop any row which has at least one missing value, by default axis is 0, so we are dropping rows, if we want to drop columns we will have to specify axis = 1
df.dropna()  
```

```python
df
```

```python
df.dropna(axis=1)   #will drop column
```

```python
#thresholding the data, will drop rows which have less than 2 non-NA values
df.dropna(thresh=2)
```

```python
df.fillna(value='FILL ')  #filling the null with a value
```

```python
#filling column A with its mean
df['A'].fillna(value=df['A'].mean())
```

# Groupby

The groupby method allows you to group rows of data together and call aggregate functions

```python
# Create dictionary
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
       'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
       'Sales':[200,120,340,124,243,350]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
#groupby Company column('df.groupby('column name')')
byComp = df.groupby('Company')
```

```python
#aggregation
byComp['Sales'].mean()
```

```python
#groupby the column Company and then aggragation performed on the Sales column
df.groupby('Company')['Sales'].mean()
```

```python
m = df.groupby('Company')['Sales'].sum()
```

```python
m
```

```python
#because of single bracket, it is a series
type(m)
```

```python
m = df.groupby('Company')[['Sales']].sum()
```

```python
print(type(m))
m
```

```python
m
```

```python
m.loc['FB']
```

```python
df.groupby('Company').max()
```

```python
df.groupby('Company').min()
```

```python
df.groupby('Company')[['Sales']].std()
```

```python
df.groupby('Company').describe()
```

```python
df.groupby('Company').describe().transpose()
```
