Plotly is a library that allows you to create interactive plots that you can use in dashboards or websites (you can save them as html files or static images).


```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.io as pio
```

```python
# renderer for VS Code notebooks
pio.renderers.default = 'vscode'
```

```python
#creating fake data
df = pd.DataFrame(np.random.randn(100,4), columns = 'A B C D'.split())
```

```python
df.head()
```

```python
#dictionary creating
d = {'Category':['A','B','C'], 'Values':[32,43,50]}
```

```python
d['Values']
```

```python
df2 =pd.DataFrame(d)
```

```python
df2
```

# Scatter

```python
# scatter plot
fig = px.scatter(
    df,
    x='A',
    y='B',
    title='Scatter Plot of A vs B'
)

# show plot
fig.show()
```

## Bar Plots

```python
fig = px.bar(df2, x= 'Category',y = 'Values', title='Bar Plot')
fig.show()
```

```python
px.bar(df2, x='Category', y='Values')
```

```python
px.bar(df.count())
```

## Boxplots

```python
px.box(df)
```

# 3D Surface

```python
#creating a dataframe with 3 features
df3 = pd.DataFrame({'x':[1,2,3,4,5],'y':[10,20,30,20,10],'z':[5,4,3,2,1]})
```

```python
import plotly.graph_objects as go

fig = go.Figure(data=[go.Surface(z=df3.values,
                                 colorscale='RdYlBu')])

fig.show()
```
